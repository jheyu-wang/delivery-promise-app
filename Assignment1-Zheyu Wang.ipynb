```python
# 1.a

def late_percentage(zones, time_blocks, promise):
    late_order = 0
    total_order= 0

    #If a particular zone and time block is provided, it returns the percentage of late orders in that zone and time block.
    if zones != "all" and time_blocks!= "all":
      #late_order = len(delivery_times(zones, time_blocks, promise) > promise)
      #total_order = len(delivery_times(zones, time_blocks, promise))

      array = delivery_times(zones, time_blocks, promise)
      late_order = np.sum(array > promise)
      total_order = len(array)

      return late_order / total_order * 100

    #If the zone is 'all' and a specific time block is provided, it returns the percentage of late order in that time block across all zones.
    elif zones == "all" and time_blocks != "all":
      for z in ZONES:
        array = delivery_times(z, time_blocks,promise)
        late_order += np.sum(array > promise)
        total_order += len(array)
      return late_order / total_order * 100
    # If a specific zone is provided and the time block is 'all', it returns the percentage of late orders in that zone across all time blocks.
    elif zones != "all" and time_blocks == "all":
      for t in TIME_BLOCKS:
        array = delivery_times(zones, t, promise)
        late_order += np.sum(array > promise)
        total_order += len(array)
      return late_order / total_order * 100


    # all zones, all times
    else:
      for z in ZONES:
        for t in TIME_BLOCKS:
          array = delivery_times(z, t, promise)
          late_order += np.sum(array > promise)
          total_order += len(array)
      return late_order / total_order * 100

```

```python
# b

rates = []

# Calculate all 12 late arrival rates
for z in ZONES:
    for t in TIME_BLOCKS:
        rate = late_percentage(z, t, 45)
        rates.append([z, t, rate])

# Get the percentages
percentages = []

for row in rates:
    percentages.append(row[2])

# Sort from highest to lowest
order = np.argsort(percentages)[::-1]

# Print the ranking
for i in order:
    print(rates[i][0], rates[i][1], round(rates[i][2], 2), "%")
```

```python
def Average_delivery_time(zones, time_blocks, promise):
  # frirst
  if zones != "all" and time_blocks != "all":
    array = delivery_times(zones, time_blocks, promise)
    return np.mean(array)

  # If the zone is 'all' and a specific time block is provided
  elif zones == "all" and time_blocks != "all":
    arrays = []
    for z in ZONES:
      arrays.append(delivery_times(z, time_blocks, promise))
    combined = np.concatenate(arrays)
    return np.mean(combined)

  # If a specific zone is provided and the time block is 'all'
  elif zones != "all" and time_blocks == "all":
    arrays = []
    for t in TIME_BLOCKS:
      arrays.append(delivery_times(zones, t, promise))
    combined = np.concatenate(arrays)
    return np.mean(combined)

  # All zones, all times
  else:
    arrays = []
    for z in ZONES:
      for t in TIME_BLOCKS:
        arrays.append(delivery_times(z, t, promise))
    combined = np.concatenate(arrays)
    return np.mean(combined)
```

```python
# testing :
print(Average_delivery_time("Far West", "Lunch", 45))

print(Average_delivery_time("all", "Lunch", 45))

print(Average_delivery_time("Far West", "all", 45))
print(Average_delivery_time("all", "all", 45))
```

```python
from IPython.core.history import DummyDB
#b

# range choosing : it cannot be too short becuase delivery takes time, for uber driver for example, 30 mph canot make huge difference within 2 min, so I chose 3 minutes
#also , it doesnt hurt if I just start from 3 min to the 60 which is way too much for a promise in general.

promise_B = list(range(3, 66, 3))
def best_promise_time(zones, time_blocks, promise_B, costs):
  profits = []
  for p in promise_B:
    array = delivery_times(zones, time_blocks, p)
    late_order = np.sum(array > p)
    total_order = len(array)

    order_profit = total_order * costs['margin'] - late_order * (costs["refund"] + costs["churn_orders"] * costs["margin"])
    profits.append(order_profit)
  best_index = np.argmax(profits)
  return promise_B[best_index], profits[best_index]
```
